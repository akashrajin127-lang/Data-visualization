```python
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for col in categorical_cols:
  df[col] = df[col].str.strip().str.title()
print(df[categorical_cols].head())
```

```python
df['average_score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
) / 3
```

```python
df['total_score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
)
```

```python
subjects = ['math score', 'reading score', 'writing score']
for col in subjects:
    print("\n", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Standard Deviation:", df[col].std())
```

```python
statistics = df[subjects].agg(['mean', 'median', 'std'])
print(statistics)
```
